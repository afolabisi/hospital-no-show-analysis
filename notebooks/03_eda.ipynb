```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the clean data
Hd = pd.read_csv('appointments_cleansed.csv')

# Create binary target for easier analysis
Hd['No_Show_Binary'] = Hd['No_Show'].map({'Yes':1, 'No':0})
```

```python
Hd['No_Show'].value_counts(normalize= True).round(3)
```

```python
Hd.groupby('Gender')['No_Show_Binary'].mean().round(3)
```

```python
Hd.groupby('Age_Group')['No_Show_Binary'].mean().round(3)
```

```python
Hd.groupby('Timing')['No_Show_Binary'].mean().round(3)
```
